```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Simulate your survey data responses (Matching Pages 40-44 of your report)
np.random.seed(42)
data_size = 150

survey_data = pd.DataFrame({
    'Age_Group': np.random.choice(['Below 18', '18–25', '26–35', '36–45', 'Above 45'], data_size, p=[0.1, 0.5, 0.2, 0.1, 0.1]),
    'Gender': np.random.choice(['Male', 'Female'], data_size),
    'Occupation': np.random.choice(['Student', 'Employed', 'Self-employed', 'Business'], data_size, p=[0.4, 0.4, 0.1, 0.1]),
    'Most_Used_FinTech': np.random.choice(['Digital wallets', 'Mobile banking', 'Investment apps', 'Loan apps'], data_size, p=[0.5, 0.3, 0.1, 0.1]),
    'Trust_Level': np.random.choice([1, 2, 3, 4, 5], data_size, p=[0.1, 0.1, 0.2, 0.4, 0.2]), # 5=Very High, 1=Very Low
    'Security_Concern': np.random.choice(['Data privacy', 'Cyber fraud', 'Technical issues', 'No concern'], data_size, p=[0.4, 0.3, 0.2, 0.1])
})

# Display the first 5 rows of your newly structured data
print("--- Cleaned Survey Data Sample ---")
print(survey_data.head())

# 2. Generate Figure 1: Age Group Distribution (From your report index)
plt.figure(figsize=(8, 5))
sns.countplot(data=survey_data, x='Age_Group', palette='Blues_r', order=['Below 18', '18–25', '26–35', '36–45', 'Above 45'])
plt.title('Figure 1: Age Group Distribution of Survey Respondents')
plt.xlabel('Age Group')
plt.ylabel('Count of Respondents')
plt.tight_layout()
plt.savefig('fig1_age_distribution.png')
plt.show()

# 3. Generate Figure 16: Security Concerns vs Age Group (From your report index)
plt.figure(figsize=(10, 6))
sns.countplot(data=survey_data, x='Age_Group', hue='Security_Concern', palette='Set2', order=['Below 18', '18–25', '26–35', '36–45', 'Above 45'])
plt.title('Figure 16: Security Concerns Across Different Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Count of Responses')
plt.legend(title='Primary Concern', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('fig16_security_concerns.png')
plt.show()
```
